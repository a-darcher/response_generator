```python
import numpy as np

import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng()
u = rng.uniform(size=1000)
t =  np.log(u)

scale = 3
threshold = 1

plt.scatter(np.arange(len(u)), u)
plt.show()

plt.scatter(np.arange(len(u)), np.log(u))
plt.show()

plt.hist(u, bins=25)
plt.show()

plt.hist(np.log(u), bins=25)
plt.show()

plt.hist(scale * np.log(u), bins=25)
plt.show()

plt.hist(threshold - scale * np.log(u), bins=25)
plt.show()
```
