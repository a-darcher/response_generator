```python
parent_path = Path("/home/al/Documents/code/generate_responses/configs/main_with_bursts")

line_number = 95         # 1-based line number


old_string =  "15"
new_string =  "15\nburst_rate_factor: 0.25"

for p in parent_path.rglob("*"):

    # Read file
    with p.open("r", encoding="utf-8") as f:
        lines = f.readlines()

    # Modify specific line
    idx = line_number - 1
    if old_string in lines[idx]:
        lines[idx] = lines[idx].replace(old_string, new_string)

    # Write back
    with p.open("w", encoding="utf-8") as f:
        f.writelines(lines)
```

```python
response_type = "baseline"


parent_path = Path("/home/al/Documents/code/generate_responses/configs/main_dataset")
p = parent_path / "example.yaml"

for trials in trial_list:

    fname = f"[{trials[0]},{trials[1]}]_{response_type}.yaml"
    save = parent_path / fname

    line_number = 16            # 1-based line number
    old_value = "[]"
    new_value = f"[{trials[0]},{trials[1]}]"

    old_string = str(old_value)
    new_string = str(new_value)


    # Read file
    with p.open("r", encoding="utf-8") as f:
        lines = f.readlines()

    # Modify specific line
    idx = line_number - 1
    if old_string in lines[idx]:
        lines[idx] = lines[idx].replace(old_string, new_string)

    # Write back
    with save.open("w", encoding="utf-8") as f:
        f.writelines(lines)
```
